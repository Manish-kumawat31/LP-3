```python

# Gradient Descent Algorithm

def function(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)

x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    x = x - learning_rate * derivative(x)

print("Minimum value of x:", x)
print("Minimum value of y:", function(x))
```
